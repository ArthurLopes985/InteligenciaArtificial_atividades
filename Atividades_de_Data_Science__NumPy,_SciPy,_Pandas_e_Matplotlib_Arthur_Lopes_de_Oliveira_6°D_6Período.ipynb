```python
import scipy.optimize as spo
def funcao(x):
  return x**4 - 3*3+2
raizes = spo.root_scalar(funcao, x0=3, method='secant')
print(f"A raiz encontrada é: {raizes.root}")
```

#4. Use a função de otimização de SciPy para maximizar f(x) = −x^2 + 4x + 1.

```python
from scipy.optimize import minimize
def funcao_invertida(x):
    valor = -x[0]**2 + 4*x[0] + 1
    return -valor
resultado = minimize(funcao_invertida, x0=[0])
x_maximo = resultado.x[0]
y_maximo = -resultado.fun
print("Ponto x onde atinge o máximo:", x_maximo)
print("Valor máximo da função (y):", y_maximo)
```

#11. Encontre os autovalores e autovetores de uma matriz 4x4 de números inteiros aleatórios.

#14. Utilize a função de otimização multivariada para minimizar f(x,y) = (x−2)^2 + (y−3)^2.

```python
from scipy.optimize import minimize
def funcao(ponto):
    x = ponto[0]
    y = ponto[1]
    return (x - 2)**2 + (y - 3)**2
resultado = minimize(funcao, x0=[0, 0])
print("Ponto mínimo encontrado:")
print("x =", resultado.x[0])
print("y =", resultado.x[1])
```

# 8. Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função y = sin(x).

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)
y = np.sin(x)
erro = 0.2
plt.plot(x, y, label='sin(x)')
plt.fill_between(x, y - erro, y + erro, alpha=0.3)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Função seno com intervalo de confiança')
plt.legend()
plt.show()
```

# 18. Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.
