#18. Número Primo

```python

num = int(input("Digite um número: "))#pede e captura o numero desejado
if num <= 1:#verifica se o numero é 1 ou menor, se for nao é primo
    print(f"{num} não é primo")
elif num <= 3:#verfica se é 2 ou 3, se for é primp
    print(f"{num} é primo")
elif num % 2 == 0 or num % 3 == 0:#se for divisivel por 2 ou por 3 não é primo
    print(f"{num} não é primo")
else: #testa divisores de 5 até a raiz quadrada do numero digitado
    i = 5
    primo = True  # Flag para mensagem de saida apos o loop
    while i * i <= num: #cria loop
        if num % i == 0 or num % (i + 2) == 0: #testa se o numero é divisivel por i ou por i+2, se for não é primo
            primo = False
            break#encerra o loop
        i += 6#incrementa mais um em i para testar novamente se nao caiu no primeiro teste
    if primo:#mensagem de acordo com a flag do loop
        print(f"{num} é primo") #se true, mensagem de primo
    else:
        print(f"{num} não é primo")#se false, mensagem de não primo
```

pede numeero, tenta dividir por alguns numeros como 2 ou 3 ou a cima de 5, se apos todos esses testes o numero só for divisivel por ele e por 1, é primo

#21.Busca Binária

```python

lista = [] #criacao da lista vazia para inserir numeros
qtd = int(input("Quantos numeros deseja inserir? ")) #pergunta pro usuario quantas vao ser inseridos, para saber quantas vezes o loop a seguir vai se repetir
for i in range(qtd): #cria loop
  num = int(input(f"Digite o {i+1}°numero: ")) #captura o numero digitado em uma variavel
  lista.append(num)#adiciona o numero na lista
lista.sort()#ordena a lista do menor ao maior
inicio = 0 #indice inicial da lista
fim = len(lista)-1 #indice final, -1 porque o indice começa em 0
num2 = int(input("Qual numero deseja buscar? ")) #pergunta pro usuario qual numero deve ser buscado
while True: #loop
        meio = (inicio + fim) // 2  # descobre o índice do meio
        if lista[meio] == num2: #verifica se o numero do meio é o buscado
            indice_achado = meio  #pega o indice achado
            break
        elif lista[meio] < num2:#caso o meio seja menor
            inicio = meio + 1  # avanca uma casa para procurar
        else:#caso o meio seja maior
            fim = meio - 1  # reduz uma casa para procurar
print(f"O indice do numero procurado é {indice_achado}")#printando a casa em que foi achada o numero
```

#22. Criptografia de Cifra de César

```python

mensagem = input("Digite a mensagem que quer criptografar: ")#pede e captura a mensagem inicial
deslocamento = int(input("Quantos de deslocamento voce quer? "))#pede e captura o deslocamento desejado
resultado = "" #inicializa variavel para o resultado final
for caractere in mensagem: #passa pelos caracteres da mensagem
  if caractere.isalpha():#verifica se é alfabetico
    base = ord('A') if caractere.isupper() else ord('a') # desloca dentro do intervalo de 26 letras
    resultado += chr((ord(caractere) - base + deslocamento) % 26 + base)#passa para o resultado ja criptografado
  else:#o que nao for letra, ou seja numeros, espaco, pontuacao, vem para essa condição
    resultado += caractere  # mantém espaços, números, pontuação
print(f"O resultado da frase criptografada é: {resultado}") #mostra o resultado ao usuario
```

pede a frase, o deslocamento, desloca as letras e printa o resultado final

#24.Validação de CPF

```python

def valida_cpf(cpf): #inicializa funcao, da o nome, atributos necessarios
  if not cpf.isdigit(): #verifica se o cpf tem apenas numeros
    return "CPF deve conter apenas numeros, CPF invalido" #se houver algo que nao seja numero, retorna mensagem de erro
  elif len(cpf) != 11: #verifica se o tamanho do cpf é 11 digitos
    return "Tamanho incorreto, CPF invalido" #se houver menos ou mais de 11 digitos, retona mensagem de erro
  else: #de acordo com a regra oficial do cpf, os ultimos dois digitos são verificadores, e são feitos atraves de operações matematicas
    for i in range(9, 11):
        soma = sum(int(cpf[num]) * ((i + 1) - num) for num in range(0, i)) #multiplica cada um dos 9 iniciais por pesos que vao de 10 a 2 e soma tudo
        digito = ((soma * 10) % 11) % 10 #multiplica por 10, pega o resto de divisão por 11, depois por 10 e esse é o digito verificador.
        if digito != int(cpf[i]):#verifica se o digito da conta é igual ao digitado
            return "Erro ao validar pelos digitos verificadores, CPF invalido"#se nao for, mensagem de erro
# Não conhecia essa regra de negocio do cpf, decobri ao fazer esse exercicio, pesquisando no google quais verificações seriam interessantes.
    return f"o cpf {cpf} é valido"#se nao cair em nenhuma validacao, mensagem de sucesso
cpf = input("Digite o CPF")#pede e captura o cpf
print(valida_cpf(cpf))#chama funcao, passa o cpf e printa a mensagem
```

#25. Classificação de Números

```python

lista = [] #criacao da lista vazia para inserir numeros
positivos = [] #criacao de lista vazia para categoria positiva
negativos = [] #criacao de lista vazia para categoria negativa
zeros= [] #criacao de lista vazia para categoria zero
qtd = int(input("Quantos numeros deseja inserir? ")) #pergunta pro usuario quantas vao ser inseridos, para saber quantas vezes o loop a seguir vai se repetir
for i in range(qtd): #cria loop
  num = int(input(f"Digite o {i+1}°numero: ")) #captura o numero digitado em uma variavel
  lista.append(num)#adiciona o numero na lista
for i in lista: #passa por todos itens da lista
  if i == 0: #verifica se é 0
    zeros.append(i)#se for 0, insere na lista de 0
  elif i > 0: #verifica se é maior que 0
    positivos.append(i)#se for maior q zero, insere na lista de positivo
  else:
    negativos.append(i)#se nao caiu em nenhum anterior, insere na lista de 0
print(f'''Lista Original: {lista}
      Positivos: {positivos}
      Negativos: {negativos}
      Zeros: {zeros}''')#print com a lista original e separados em categorias
```

#30.Jogo da Forca

```python

import random #importa biblioteca aleotoria
palavras = ["verstappen", "leclerc", "hamilton", "antonelli", "bortoleto", "russell", "norris", "piastri", "hadjar", "hulkenberg"] #lista de palavras possiveis
palavra_forca = random.choice(palavras)#sorteia uma palavra na lista de opções
letras_certas = ["_"] * len(palavra_forca) #lista para acumular letras corretas
letras_erradas = [] #lista para acumular as letras erradas
tentativas = 6 #numero de tentativas
print("Adivinhe a palavra:") #print introdutório
print(" ".join(letras_certas))
while tentativas > 0 and "_" in letras_certas: #loopp enquanto ainda tem tentativas e faltam letras
      tentativa = input("Digite uma letra: ").lower() #pede e captura tentativa de letra
      if not tentativa.isalpha() or len(tentativa) != 1: #verificacao se só tem uma letra digitada
          print("Digite apenas uma letra.")
          continue
      if tentativa in letras_certas or tentativa in letras_erradas: #verificacao se a letra não é reptida
          print("Essa letra já foi.")
          continue
      if tentativa in palavra_forca: #condições se a letra e correta, colocando ela no lugar
          for i, letra in enumerate(palavra_forca):
              if letra == tentativa:
                    letras_certas[i] = letra
                    print("Letra encontrada.")
      else:#condicao se a letra for errada, tirando uma tentativa
          letras_erradas.append(tentativa)
          tentativas -= 1
          print(f"Letra errada! Faltam: {tentativas} chances")
      print(" ".join(letras_certas))
      print(f"Letras erradas: {', '.join(letras_erradas)}")
if "_" not in letras_certas:#se nao houver espacos vazios, mensagem de vitória
    print(" Você acertou a palavra!")
else:#se nao houver mais tentativas, aparece mensagem de derrota
    print(f"Morreu. A palavra era: {palavra_forca}")
```

importa biblioteca aleatoria, cria lista com pilotos de f1, sorteia um deles, pede letra, se tiver aparece o local, se nao entra em erro, acaba ao descobrir a palavra ou 6 erros.

```python

tab = [
    ["", "", ""],
    ["", "", ""],
    ["", "", ""]
    ]#inicializa o tabuleiro
jogador = "X" #inicializa com o jogador X
print("Bem vindo ao jogo da velha!")#Mensagem introdutória
def tabuleiro():
   for i in range(3): #percorre as linhas do tabuleiro
      print("|".join(tab[i]))#junta os elemntros da linha separando com |
      if i<2:#confere se nao é a ultima linha, para não printar separador nela
        print("-"*5)
tabuleiro()
while True:
  linha = int(input(f"Jogador {jogador}, escolha a linha (0, 1, 2): "))
  coluna = int(input(f"Jogador {jogador}, escolha a coluna (0, 1, 2): "))#pede a linha e a coluna desejada pelo usuario
  if linha < 0 or linha > 2 or coluna < 0 or coluna > 2:
    print("Entrada invalida")#checa se o usuario nao digitou nada de errado
    continue
  if tab[linha][coluna] !="": #se a linha e coluna digitada ja tiver ido, printa mensagem de erro
    print("Esse lugar já foi! Tente novamente!")
    continue
  tab[linha][coluna] = jogador #faz a jogada
  vencedor = False#iniciaiza variavel para saber se venceu
  for linha in tab:
    if all(c == jogador for c in linha): #confere os elementos por linha
      vencedor = True#troca para vencedor se der vitoria em alguma linha
  for coluna in range(3):
    if all(tab[linha][coluna] == jogador for linha in range(3)):#verifica por coluna agora
      vencedor = True #troca para vencedor se der alguma vitoria em coluna
  if all(tab[i][i] == jogador for i in range(3)): #chega diagonal 1( esq. cima para dir. baixo)
    vencedor = True#troca para vencedor se der vitoria nessa coluna
  if all(tab[i][2-i] == jogador for i in range(3)): #chega diagonal 2( dir. cima para esq. baixo)
    vencedor = True#troca para vencedor se der vitoria nessa coluna
  if vencedor == True:#se vecendor for true, aparece mensagem de vitória e encerra jogo
      print(f"Parabéns! Jogador {jogador} venceu!")
      break
  if all(c!="" for linha in tab for c in linha):#se todas casas tiverem preenchidas, e ninguem venceu ainda, da mensagem de empate e encerra o jogo
      print("Empate!")
      break
  jogador = "O" if jogador =="X" else "X" #se ninguem venceu ou nao foi empate, troca o jogador para bolinha se for x e vice-versa
  tabuleiro()
```

#50.Cálculo de Frequência de Letras

```python

frase = input("Digite a frase desejada: ") #pede e captura a frase digitada pelo usuario em uma variavel
letras = [] #lista vazia para armazenar as letras no proximo passo
contador = {} #dicionario que no final vai armazenar letras e frequencia
for letra in frase: #passa oor cada uma das letras da frase digitada pelo usuario
  letras.append(letra) #adiciona a letra na lista inicianlizada anteriormente
for l1 in letras:
    if l1 in contador: #se a letra já estiver no dicionario, incrementa em 1 na frequencia
        contador[l1] += 1
    else: #se a letra não estiver ainda, é adicionado no dicionario com frequencia = 1
        contador[l1] = 1
print(contador) #printando cada letra detectada e sua respectiva frequencia
```
